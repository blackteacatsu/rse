## Q9: Name at least two potential sources of error in your calculation.

Here are two sources of error in the NDVI change methods:

1. **Vegetation changes unrelated to fire:** NDVI responds to differences between near-infrared (NIR) and red bands rather than uniquely to burning. Seasonal vegetation changes and other environmental changes between the two observations could therefore be mistaken for fire-related NDVI decreases, causing an overestimate of burned area.

2. **Threshold and binary errors:** Using a binary classification method as Question 8 let the threshold to assign every pixel to either the burned or unburned class, even though fire effects and the proportion burned within a pixel can vary continuously. Therefore, lightly burned or mixed pixels may be classified as unbruned if their NDVI decline is too small to meet the the threshold.

## Question 14: List at least two other techniques or combination of techniques that you could have used to estimate burned area.

Here are two other approaches techniques that one could use to estimate burned area:

1. **Radar/SAR backscatter change detection:**

   Leaves and branches scatter the transmitted radar signal in multiple directions, producing canopy volume scattering, and can be detected using cross-polarized measurements such as HV or VH. If fires remove foliage and branches, the canopy contribution may decrease, while scattering from the exposed ground and remaining trunks becomes dominant.

   Compare aligned pre-fire and post-fire SAR images acquired at similar viewing angles and the same polarization to identify changes in backscatter.

   But this also relies on the assumption that no other disturbances have taken place and contribute to the loss of canopy scattering during the observation windows. Low-severity surface fires may also leave little canopy change to detect.

2. **Thermal hotspot anomaly:**

   Identify unusually hot pixels in thermal infrared imagery during the fires to locate active burning. Combining observations throughout the fire period could help track fire progression, while post-fire optical imagery could delineate the burned footprint for area calculation. Thermal detection alone may miss fires between satellite passes and not directly measure the complete burned area.

   MODIS aboard Terra and Aqua provides an example of thermal fire monitoring from near-polar, sun-synchronous Earth orbits. A geostationary instrument could provide more frequent observation of the same region, although its greater distance generally makes fine spatial resolution more difficult to achieve.
