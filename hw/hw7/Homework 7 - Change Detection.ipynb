### Q9: Name at least two potential sources of error in your calculation.

1. The NDVI is a index derived from calculates the difference between near-infrared (NIR) and red light reflectance, divided by their sum. So it also depends on the time of year and bioproductivity of the plants at the time of our observations, burned in deciduous plants whose productivity is strongly influenced by seasonality might not be accounted for during this studies.

2. One assumption we need to made during this analysis is no other major events happened between the interval of the two observations used for our comparison, so that we could eventually contribute all differences in NDVI to a single source in this case.
