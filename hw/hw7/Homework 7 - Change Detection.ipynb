
### Q9: Name at least two potential sources of error in your calculation.

Here are two sources of error in the NDVI change methods:

1. **Vegetation changes unrelated to fire**: differences in seasonal growth, senescence, or drought stress can change NDVI between observations. Unburned vegetation could therefore be classified as burned, overestimating burned area.

NDVI is a index derived from calculates the difference between near-infrared (NIR) and red light reflectance, divided by their sum. So it also depends on the time of year and bioproductivity of the plants at the time of our observations, burned in deciduous plants whose productivity is strongly influenced by seasonality might not be accounted for during this studies.

2. **Threshold and mixed-pixel errors:** a single threshold may miss lightly burned areas or pixels containing both burned and unburned vegetation, underestimating burned area. It may also include unburned pixels whose NDVI decreased for other reasons.

One assumption we need to made during this analysis is no other major events happened between the interval of the two observations used for our comparison, so that we could eventually contribute all differences in NDVI to a single source in this case.
