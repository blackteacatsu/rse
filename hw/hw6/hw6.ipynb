### Question 5 Explain the chl-a and SST values in your coastal vs. offshore points for a typical year (1998, 1999, or 2000). Why might one expect these patterns?

This might be due to equatorial upwelling zone, where near surface trade wind along away from shore, thereby warm surface water is also transported offshore. Eventually, deeper, cold water surface to replenish the displaced water, thus causing the coastal sea surface temperature to be colder than the offshore sea surface temperature.
